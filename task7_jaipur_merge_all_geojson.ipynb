```python
shapes = []
scene_ids = []

# Iterate through each feature in the shapes data
for feature_tuple in data:
    scene_id, geom_data = feature_tuple
    scene_ids.append(scene_id)  # collect scene id
    geom = {'type': geom_data['type'], 'coordinates': geom_data['coordinates']}
    s = shape(geom)
    shapes.append(s)
```

```python
# Create a single polygon object from these multiple polygons
merged = cascaded_union(shapes)

# Convert the Shapely geometry object back into GeoJSON format
merged_geojson = mapping(merged)

# Print out the scene IDs and the GeoJSON of the merged polygon
# print(f"Scene IDs: {scene_ids}")
# print(f"Merged GeoJSON: {json.dumps(merged_geojson)}")

# Convert GeoJSON to Geopandas GeoDataFrame
gdf = gpd.GeoDataFrame([{'geometry': shape(merged_geojson)}])
gdf

# Export the GeoDataFrame to a shapefile
gdf.to_file('merged_jaipur_all.shp')
```
