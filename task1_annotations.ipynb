# import packages


```python
import matplotlib.pyplot as plt
import numpy as np
```

# create plotting function

```python
def plot_mask(num):
    f= plt.imread(f'../../data/task_1_segmentation_test/images/chip{num}.jpg')
    f_label= plt.imread(f'../../data/task_1_segmentation_test/SegmentationClass/chip{num}.png')
    mask= (f_label==0)[:,:,0]
    ma_img_rgb = np.concatenate([f_label, ~mask[:, :, np.newaxis]], axis=-1)
    plt.imshow(f)
    plt.imshow(ma_img_rgb, alpha=0.2)
```

# plot data

```python
plot_mask(0)
```
